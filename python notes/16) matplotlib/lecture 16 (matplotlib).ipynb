# Scatter Plot

```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5,6]
y = [4,2,3,5,0,3]
plt.title("Scatter Plot")
plt.xlabel("month")
plt.ylabel("numbers")

c = (['red','magenta','green','orange', 'blue', 'brown'])

plt.scatter(x,y, color= c)
plt.show()
```

```python
x = [1,2,3,4,5,6]
y = [4,2,3,5,0,3]
plt.title("Scatter Plot")
plt.xlabel("month")
plt.ylabel("numbers")


plt.scatter(x,y, color= "green")
plt.show()
```

```python
x = [1,2,3,4,5,6]
y = [4,2,3,5,0,3]
plt.title("Scatter Plot")
plt.xlabel("month")
plt.ylabel("numbers")


plt.scatter(x,y, color= "green", s=500)
plt.show()
```

# matplotlib - pie

```python
x = [10,20,30,40]
y = {"English","Urdu","Hindi","Chinese"}
c = {"yellow","magenta", "aqua","orange"}
plt.pie(x, labels=y, colors = c)
plt.legend()
plt.show()
```
