# Tuples In Python

# Creating A Tuple

# Single element

```python
my_tuple = (1,)
print(my_tuple)
print(type(my_tuple))
```

# Multi element

```python
my_tuple = (1,2,3,3)
print(my_tuple)
print(type(my_tuple))
```

# Mixed Data Types

```python
my_tuple = (1,"hello",3.2,4+5j)
print(my_tuple)
print(type(my_tuple))
```

# Empty Tuple

```python
my_tuple = ()
print(my_tuple)
print(type(my_tuple))
```

# Nested Tuple

```python
a = ("data",[1,2,3],(4,5,6))
print(a)
print(type(a))
```

# Creating A Tuple with only one item

```python
var2 = ("hello",)
print(var2)
print(type(var2))
```

```python
var2 = ("5",)
print(var2)
print(type(var2))
```

# Tuple Constructor

```python
a = tuple(("a","b","c"))       //Constructor
print(a)
print(type(a))
```
