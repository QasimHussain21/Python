# While Loop

```python
count = 1
while count<=5:
    print(count)
    count = count + 1
```

# Write a python program for even numbers between 1 and 50 using a while loop

```python
num = 1
while num <=50:
    if num%2 == 0:
        print(num)
    num = num+1    
```

# While Loop, using if else

```python
email = input("Enter your mail id:")

valid_email = False

while not valid_email:
     if "@" in email and "." in email:
         print("your email id is valid")
         valid_email = True
     else:
         print("your mail id is not valid try again")
         email = input("plz enter your mail id:")
```

# For loop and if condition.

```python
for i in range(0,26):
    if i%2 == 0:
        print("the number is even", i)
    else:
        print("the number is odd", i)
```

# Implement a Python program to generate the multiplication table of a given number using a for loop.

```python
number = int(input("enter a number to generate its multiplication table:"))
print("The multiplication table of", number)

for i in range(1,11):
    print(f"{number} x {i} = {number*i}")
```

```python
num_items = int(input("enter the number of items purchased"))
total_price = 0

for i in range(num_items):
    price = float(input("enter the price of the item"))
    total_price = total_price + price
    i = i+1

if total_price >=1000:
    discount = 0.1*total_price
elif 50<= total_price <100:
     discount = 0.5*total_price
else:
    discount = 0

total_bill = total_price - discount
print("subtotal",total_bill)
```

# break statement

```python
for i in range(5):
    if i ==3:
        break
    print(i)    
```

# continue statement

```python
for i in range(5):
    if i == 3:
        continue
    print(i)
print("I am learning python")    
```
