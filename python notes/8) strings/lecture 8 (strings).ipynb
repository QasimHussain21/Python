```python
def extract_date_parts(date_string):
    day, month, year = date_string.split('-')
    return day, month, year
date_string = "24-7-2025"
day, month, year = extract_date_parts(date_string)
print("Day:", day)
print("Month:", month)
print("Year:", year)     
```
