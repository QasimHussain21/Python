# SET

```python
*sets use {}
*sets are mutable
*unordered
*unique elements(duplicate elements automatically removed)
```

# creating a set

```python
a = {"animals","cat","man"}
print(a)
print(type(a))
```

# Properties of set

# 1) Set items are unordered

# 2) Indexing not allowed

```python
a = {"animals","cat","man"}
print(a[0])
```

# Duplicates not allowed

```python
a = {"animals","cat","man","man","cat","blue"}
print(a)
```

# True and 1 is consider the same value

```python
a = {"animals","cat","man","man","cat","blue",1,2,3, True}
print(a)
```

# Get the length of a set

```python
a = {"animals","cat","man","man","cat","blue",1,2,3, True}
print(a)
print(len(a))
```

# Update Items

```python
x = {"data" , "qasim" , "muneeb"}
y = {"123", True,"qasimbhai", 2.21, 123}

x.update(y)
print(x)
```

# Remove Set Items

```python
n ={70,40,"python",True}
n.remove(True)
print(n)
```

# pop Method

```python
m ={70,40,"python",True}
x = m.pop()
print(m)
```

# Clear Method

```python
m ={70,40,"python",True}
x = m.clear()
print(m)
```

# Join Two Sets

```python
x = {1,2,True,2.3,-987.33}
y = {22,4.5,6.87,False}
z = x.union(y)
print(z)
```

# Write a python program to find the max and min values

```python
setn = {21,10,3121}
print("Priginal Set Elements")
print (setn)
print ("Maximum value of the set is")
print (max(setn))
print ("Minimum value of the set is")
print (min(setn))
```

# Return a new set of identical items from two sets

```python
set1 = {1,21,3}
set2 = {21,44,56}

print(set1.intersection(set2))
```

```python
set1 = {1,21,3}
set2 = {21,44,56}

print(set1.difference(set2))
```
