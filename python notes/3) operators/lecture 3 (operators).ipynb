# Lecture no 3 (Operators):-

# 1) ARITHMETIC OPERATOR

```python
a = 3
b = 2
print(a+b)   # add
print(a-b)   # subtract
print(a%b)   # Modulus
print(a/b)   # Division
print(a//b)  # Floor Division
print(a*b)   # Multiplication
print(a**b)  # Exponentiation or power
```

# 2) ASSIGNMENT OPERATOR

```python
a=5
b=6
a = b
print(a)
a += b
print(a)
a -= b
print(a)
a *= b
print(a)
a /= b
print(a)
a //= b
print(a)
a %= b
print(a)
a **= b
print(a)
```

# 3) Comparision Operator/ Relational operators

```python
a = 4
b= 6
print(a==b)
print(a!=b)
print(a>b)
print(a<b)
print(a>=b)
print(a<=b)
```

# 4) Logical Operator

```python
a = True
b = False
print(a and b)
print(a or b)
print(not b) 
```

# 5) Bitwise Operator

```python
a = 4
b = 43
print(a & b)
print(a | b)
print(a ^ b)
print(~ b)
print(a << b)
print(a >> b)
```

# 6) Membership Operator 

```python
a = 12
b = [4, 4 , 6]
print(a in b)
print(a not in b)
```

# 7) Identify Operator

```python
a = 12
b = 5
print(a is b) 
print(a is not b)
```

```python
num = int(input("Enter a 3 digit number: "))

d1 = num % 10
num = num // 10

d2 = num % 10
num = num // 10

d3 = num % 10

total = d1 + d2 + d3

print("Sum of digits =", total)
```
