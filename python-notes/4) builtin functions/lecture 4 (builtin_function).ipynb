```python
price = 88.8838388

# Basic formatting
print("format {}".format(price))

# Limit decimal places
print("format {:.2f}".format(price))   # → format 88.88

# Using f-strings (Python 3.6+ recommended)
print(f"format {price:.2f}")
```
